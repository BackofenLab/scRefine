```python
import matplotlib.pyplot as plt
import numpy as np

from sklearn.datasets import make_blobs

label = "toy"
```

```python
plt.figure(figsize=(8, 6))
plt.title("Two blobs", fontsize="small")
X1, Y1 = make_blobs(n_samples=30000, n_features=2, centers=4)
plt.scatter(X1[:, 0], X1[:, 1], marker="o", c=Y1, s=25, edgecolor="k")
```

```python
val_size = 0.1 # 10%
true_count = int(X1.shape[0] * val_size)
false_count = X1.shape[0] - true_count

val_flag = np.concatenate((np.full(true_count, True), np.full(false_count, False)))
np.random.shuffle(val_flag)
```

```python
np.savez(f'./data/{label}/train_toy_4blobs.npz', data=X1[~val_flag], labels=Y1[~val_flag], genes=None)
np.savez(f'./data/{label}/val_toy_4blobs.npz', data=X1[val_flag], labels=Y1[val_flag], genes=None)
```
